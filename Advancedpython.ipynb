```python
#Creating a class,object and displaying details
class student:
    def __init__(self,name,age,course,mobile):
        self.name = name
        self.age = age
        self.course = course
        self.mobile = mobile
    def display(self):
        print(f"Name:{self.name},Age:{self.age},Course:{self.course},Mobile:{self.mobile}")
    def study(self,subject):
        print(f"{self.name} is studying {subject}")
    def study2(self):
        print(f"{self.name} is studying {self.course}")
        
s = student('Pavi',22,'MCA',124)
s.display()
s.study("python")
s.study2()
```

```python
#we can create methods in three ways init,class and type
#classmethod
class circle:
    pi = 3.14
    def __init__(self,radius):
        self.radius = radius
    @classmethod
    def change_pi(cls,value):
        cls.pi = value
    def area(self):
        return circle.pi*self.radius**2
c = circle(5)
print(c.area())
circle.change_pi(6)
print(c.area())

    
```

```python
#staticmethod
class circle:
    pi = 3.14
    def __init__(self,radius):
        self.radius = radius
    @classmethod
    def change_pi(cls,value):
        cls.pi = value
    @staticmethod
    def info():
        print("This is the area of circle")
    def area(self):
        return circle.pi*self.radius**2
c = circle(5)
print(c.area())
circle.change_pi(6)
print(c.area())
c.info()
```

```python
#single inheritance
class parent:
    def display(self):
        print("This is a parent class")
class child(parent):
    def show(self):
        print("This is a child class")
c = child()
c.display()
c.show()
    
```

```python
#Multiple Inheritance
class parent:
    def display(self):
        print("This is a parent class")
class parent2:
    def display2(self):
         print("This is a second parent class")
class child(parent,parent2):
    def show(self):
        print("This is a child class")
c = child()
c.display()
c.display2()
c.show()
```

```python
#MultiLevel Inheritance
class parent:
    def display(self):
        print("This is a parent class")
class parent2(parent):
    def display2(self):
         print("This is a second parent class")
class child(parent2):
    def show(self):
        print("This is a child class")
c = child()
c.display()
c.display2()
c.show()
```

```python
#shopping program using inheritance
class product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def display1(self):
        print(f"Product Name:{self.name},\nPrice:{self.price}")
class clothing(product):
    def __init__(self,name,price,warranty):
       super().__init__(name,price)
       self.warranty = warranty
    def display2(self):
        self.display1()
        print(f"Warranty:{self.warranty} years")
c = clothing('Dress',1200,2)
c.display2()

        
```

```python
#multilevel inheritance
class emp:
    def __init__(self,id,name):
        self.id = id
        self.name = name
    def display1(self):
        print(f"Employee id:{self.id},\nEmployee Name:{self.name}")
class pgm(emp):
    def __init__(self,id,name,language):
       super().__init__(id,name)
       self.language= language
    def display2(self):
        self.display1()
        print(f"Programming Language:{self.language}")
class size(pgm):
    def __init__(self,id,name,language,teamsize):
        super().__init__(id,name,language)
        self.teamsize = teamsize
    def display3(self):
        self.display2()
        print(f"Teamsize:{self.teamsize}")
c = size(12,'Pavi','python',3)
c.display3()
```

```python
#multiple Inheritance
class phone:
    def __init__(self,brand):
        self.brand = brand
    def display(self):
        print(f'Brand: {self.brand}')
class image:
    def show(self):
        print(f'capturing image with {self.brand} mobile')
class calling(phone, image):
    def __init__(self, brand):
        phone.__init__(self,brand)
    def say(self):
        self.display()
        self.show()
        print(f'calling with: {self.brand}')
c = calling('Iphone')
c.say()
```

```python
#Encapsulation - private variables
class student:
    def __init__(self):
        self.name = "Pavi"
        self.__age= 20
    def display(self):
        print('age',self.__age)
s = student()
print(s.name)
s.display()
```

```python
class bank:
    def __init__(self,name,balance):
        self.name = name
        self.__balance = balance
    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
    def withdraw(self,amount):
        if amount > self.__balance:
            print("Insufficient Balance")
        else:
            self.__balance -= amount
    def get_balance(self):
        return self.__balance
acc = bank("Pavi",1000)
acc.deposit(5000)
acc.withdraw(1000)
print(acc.get_balance())
```
