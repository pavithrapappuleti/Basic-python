```python
#Program1(a) : Printing power using lambda
A = [10,20,30,67]
res1 = list(map(lambda c :c**2,A ))
print(*res1)
```

```python
#Program1(b) : Finding even using filter in lambda
A = [10,20,30,67]
res = list(filter(lambda c :c%2==0,A ))
print(*res)
```

```python
#Program1(c) : Printing factorials 
def fact(n):
    if n==1:
        return 1
    else:
       return n*fact(n-1)
print(fact(5))
```

```python
#Program 2 : Fibonacci series
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)

n = int(input("Enter the number: "))
for i in range(n):
    print(fib(i),end=" ")
```

```python
#Program 10 : Exception Handling
try:
    num = int(input())
    result = 100/num
    print(result)
except ZeroDivisionError:
    print("Error : Cannot divide by zero")
except ValueError:
    print("Enter a valid number")

else:
    print("Succesfully Executed :",result)
finally:
    print("Execution Completed")
```
