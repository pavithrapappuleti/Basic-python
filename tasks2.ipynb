```python
#3.Write a Python program to implement inheritance where a class Shape has a method area(), and subclasses Rectangle and Circle override it to calculate their respective areas.  
class Shape:
    def area(self):
        print("Area of a shape")
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        return self.length * self.breadth
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14*self.radius*self.radius
r = Rectangle(3,4)
print(r.area())
c = Circle(3)
print(c.area())
```

```python
#7.  Write a Python program to create a custom iterator class that generates the squares of numbers from 1 to N using __iter__() and __next__().  
class Squares:
    def __init__(self, n):
        self.n = n
        self.num = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.num <= self.n:
            square = self.num ** 2
            self.num += 1
            return square
        else:
            raise StopIteration
s = Squares(5)
for i in s:
    print(i,end=" ")
```

```python
# 8.Write a Python program using a generator function to produce an infinite sequence of even numbers, and print the first 10 values from it.  
def even_numbers():
    num = 2
    while True:
        yield num
        num += 2
gen = even_numbers()
for i in range(10):
    print(next(gen),end=" ")
```

```python
#9. Write a Python program to create a decorator that logs the execution time taken by any function it wraps.
import time
def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print("Execution Time:", end - start, "seconds")
    return wrapper

@timer
def display():
    for i in range(1000000):
        pass
    print("Function Executed")

display()
```

```python
#10.Write a Python program to create a decorator that checks whether the arguments passed to a function are positive numbers before allowing the function to execute.  

def check_positive(func):
    def wrapper(a, b):
        if a > 0 and b > 0:
            return func(a, b)
        else:
            print("Error: Arguments must be positive numbers.")
    return wrapper

@check_positive
def add(x, y):
    print("Sum:", x + y)
add(10, 20)
add(-5, 10)
```

```python
#13. Write a Python program using the datetime module to calculate a person's age in years, given their date of birth.  
from datetime import datetime

dob = input("Enter your date of birth (DD-MM-YYYY): ")

birth_date = datetime.strptime(dob, "%d-%m-%Y")

today = datetime.today()

age = today.year - birth_date.year

if (today.month, today.day) < (birth_date.month, birth_date.day):
    age -= 1
print("Age:", age, "years")
```

```python
#16.Write a Python program to implement memoized (cached) recursion to efficiently calculate the Nth Fibonacci number for large values of N
memo = {}
def fibonacci(n):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacci(n - 1) + fibonacci(n - 2)
    return memo[n]

n = int(input("Enter N: "))
print("Fibonacci number:", fibonacci(n))
```

```python
#17.Write a Python program to create a dictionary of products and their prices, then find and display the three most expensive products.
products = {
    "Laptop": 60000,
    "Mobile": 30000,
    "Tablet": 25000,
    "TV": 50000,
    "Headphones": 5000,
    "Camera": 45000
}
top_products = sorted(products.items(), key=lambda x: x[1], reverse=True)
print("Three Most Expensive Products:")
for product, price in top_products[:3]:
    print(product, ":", price)
```

```python
#18.Write a Python program to take a set of numbers from the user and perform union, intersection, and difference operations between two sets.  

set1 = set(map(int, input().split()))
set2 = set(map(int, input().split()))

# Union
print("Union:", set1 | set2)

# Intersection
print("Intersection:", set1 & set2)

# Difference
print("Difference (Set1 - Set2):", set1 - set2)
print("Difference (Set2 - Set1):", set2 - set1)
```

```python
#20. Write a Python program to create a custom exception class InvalidAgeError, and use it in a function that validates whether an entered age is between 0 and 120, raising the custom exception otherwise.
class InvalidAgeError(Exception):
    pass

age = int(input("Enter age: "))

try:
    if age < 0 or age > 120:
        raise InvalidAgeError
    else:
        print("Valid Age")

except InvalidAgeError:
    print("Invalid Age! Age should be between 0 and 120.")
```
